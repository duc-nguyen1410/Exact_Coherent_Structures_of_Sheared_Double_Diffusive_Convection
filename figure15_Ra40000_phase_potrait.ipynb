## Compare mean profiles of DNS and ECS

```python
from func import *
def load_data(path):
    import pandas as pd
    df = pd.read_csv(path, delimiter=', ', header=None, names=['T', 'S', 'Dens'])
    mean_T = df['T'].values
    mean_S = df['S'].values
    mean_Dens = df['Dens'].values
    return mean_T,mean_S,mean_Dens

def load_mean_data_PO(path):
    file_list = getpath(path)
    data = []
    for file in file_list:
        [y,xz_ave_T] = xz_averaged_salar(file)
        data.append(xz_ave_T)
    return np.mean(data,axis=0)



plt.figure(figsize=(3, 3))

[y,xz_ave_T] = xz_averaged_salar('Data/Ra40000/E1_up/tbest.nc')
plt.plot(xz_ave_T-y,y,color='r', linestyle = 'dashed')
[y,xz_ave_T] = xz_averaged_salar('Data/Ra40000/E2_up/tbest.nc')
plt.plot(xz_ave_T-y,y,color='g', linestyle = 'dashed')
[y,xz_ave_T] = xz_averaged_salar('Data/Ra40000/E3_up/tbest.nc')
plt.plot(xz_ave_T-y,y,color='k', linestyle = 'dashed',label=r'$\bar{T}(y)$')

[T_dns,S_dns,_] = load_data('Data/Ra40000/dns/mean_profiles_dns.csv')
plt.plot(T_dns,y[1:],color='tab:blue', linestyle = 'solid', linewidth = 3, zorder = 0)

T_PO1 = load_mean_data_PO('Data/Ra40000/PO1/t*.nc')
plt.plot(T_PO1-y,y,color='tab:brown', linestyle = 'dashed')

T_PO2 = load_mean_data_PO('Data/Ra40000/PO2/t*.nc')
plt.plot(T_PO2-y,y,color='m', linestyle = 'dashed')

plt.xlabel(r'$\langle T \rangle_h$', fontsize=16)
plt.ylabel(r'$z$', fontsize=16)
plt.xticks([-0.5, 0, 0.5],['$-0.5$','0','0.5'],fontsize=16)
plt.yticks([-0.5, 0, 0.5],['$-0.5$','0','0.5'],fontsize=16)
plt.gca().set_aspect('equal')
plt.grid()
plt.savefig('Data/Ra40000/MeanProfiles_temp.png', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```
