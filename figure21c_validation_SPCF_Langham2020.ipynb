```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
# Set the default font to Times New Roman
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman']
plt.rcParams['mathtext.fontset'] = 'stix'  # Use STIX fonts for math symbols
plt.rcParams['text.usetex'] = True # enable LaTeX rendering

df = pd.read_csv("Data/Validations/Validation_StratifiedPCF_Langlam2020/figure2a_langlamjfm2020_Pr1e-4.csv", sep=', ')
array = df.values
Ri_1En4 = array[:,0]
WallShear_1En4 = array[:,1]

df = pd.read_csv("Data/Validations/Validation_StratifiedPCF_Langlam2020/figure2a_langlamjfm2020_Pr1e-2.csv", sep=', ')
array = df.values
Ri_1En2 = array[:,0]
WallShear_1En2 = array[:,1]

df = pd.read_csv("Data/Validations/Validation_StratifiedPCF_Langlam2020/figure2a_langlamjfm2020_Pr1.csv", sep=', ')
array = df.values
Ri_1 = array[:,0]
WallShear_1 = array[:,1]

df = pd.read_csv("Data/Validations/Validation_StratifiedPCF_Langlam2020/figure2b_langlamjfm2020_Pr7.csv", sep=', ')
array = df.values
Ri_7 = array[:,0]
WallShear_7 = array[:,1]

df = pd.read_csv("Data/Validations/Validation_StratifiedPCF_Langlam2020/Pr1/MuE.asc", sep='\s+')
array = df.values
Ri_1_ddc = np.append(0,array[:,0])
WallShear_1_ddc = np.append(1.19247,array[:,6])-1

df = pd.read_csv("Data/Validations/Validation_StratifiedPCF_Langlam2020/Pr1e-2/MuE.asc", sep='\s+')
array = df.values
Ri_1En2_ddc = array[:,0]
WallShear_1En2_ddc = array[:,6]-1

df = pd.read_csv("Data/Validations/Validation_StratifiedPCF_Langlam2020/Pr1e-4/MuE.asc", sep='\s+')
array = df.values
Ri_1En4_ddc = array[:,0]
WallShear_1En4_ddc = array[:,6]-1

df = pd.read_csv("Data/Validations/Validation_StratifiedPCF_Langlam2020/Pr7/MuE.asc", sep='\s+')
array = df.values
Ri_7_ddc = array[:,0]
WallShear_7_ddc = array[:,6]-1

fig = plt.figure(figsize=(3, 3))

plt.plot(Ri_1_ddc,WallShear_1_ddc,color='k', zorder=1, label=r'ChFlow--DDC')
plt.plot(Ri_1En2_ddc,WallShear_1En2_ddc,color='tab:blue', zorder=1)
plt.plot(Ri_1En4_ddc,WallShear_1En4_ddc,color='tab:orange', zorder=1)
plt.plot(Ri_7_ddc,WallShear_7_ddc,color='tab:green', zorder=1)
plt.scatter(Ri_1En4,WallShear_1En4,facecolor="none",alpha=1,color='tab:orange')
plt.scatter(Ri_1En2,WallShear_1En2,facecolor="none",alpha=1,color='tab:blue')
plt.scatter(Ri_1,WallShear_1,facecolor="none",alpha=1,color='k', label=r'Langham \textit{et al.} (2020)')
plt.scatter(Ri_7,WallShear_7,facecolor="none",alpha=1,color='tab:green')
plt.xlabel(r'$Ri_b$', fontsize=14)
plt.ylabel(r'$\tau_w-1$', fontsize=14)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.xlim([1e-2, 140])
plt.ylim([0, 6])
plt.semilogx()
plt.legend()

plt.savefig('Data/Validations/Validation_StratifiedPCF_Langlam2020/para_cont_Langham2020.png', dpi=300, bbox_inches='tight')
plt.show()
```
