# Building linear system and matrix
First, we have the governing equations as follows:
\begin{equation}
\nabla\cdot \vec{u}_{tot} = \nabla\cdot (u_b\vec{e}_x + \vec{u}) = \nabla\cdot \vec{u} = 0,
\end{equation}

\begin{equation}
\partial_t \vec{u}+u_b\partial_x\vec{u} + w\partial_zu_b\vec{e}_x + \vec{u}\cdot\nabla\vec{u} = -\nabla p + \sqrt{\frac{Pr}{Ra}}\nabla^2\vec{u} + (T-\Lambda S)\vec{e}_z,
\end{equation}

\begin{equation}
\partial_t \theta +u_b\partial_x \theta + \vec{u}\cdot\nabla \theta - w = \frac{1}{\sqrt{PrRa}}\nabla^2 \theta,
\end{equation}

\begin{equation}
\partial_t s +u_b\partial_x s + \vec{u}\cdot\nabla s - w = \frac{1}{Le\sqrt{PrRa}}\nabla^2 s.
\end{equation}

In these equations, there is presence of nonlinear terms as $\vec{u}\cdot\nabla\vec{u}$, $\vec{u}\cdot\nabla \theta$, and $\vec{u}\cdot\nabla s$. To implement the linear stability analysis, these nonlinear terms would be removed from the system. Now, the system is replace by
\begin{equation}
\nabla\cdot \vec{u} = 0,
\end{equation}

\begin{equation}
\partial_t \vec{u}+u_b\partial_x\vec{u} + w\partial_z u_b\vec{e}_x = -\nabla p + \sqrt{\frac{Pr}{Ra}}\nabla^2\vec{u} + (T-\Lambda S)\vec{e}_z,
\end{equation}

\begin{equation}
\partial_t \theta +u_b\partial_x \theta + \vec{u}\cdot\nabla \theta - w = \frac{1}{\sqrt{PrRa}}\nabla^2 \theta,
\end{equation}

\begin{equation}
\partial_t s +u_b\partial_x s + \vec{u}\cdot\nabla s - w = \frac{1}{Le\sqrt{PrRa}}\nabla^2 s.
\end{equation}

\begin{align}
\boldsymbol{\nabla}\cdot \boldsymbol{u} &= 0,\\
\frac{\partial u}{\partial t} &= -u_b\frac{\partial u}{\partial x} - w\partial_z u_b -\frac{\partial p}{\partial x} + \sqrt{\frac{Pr}{Ra}}\nabla^2 u,\\
\frac{\partial v}{\partial t} &= -u_b\frac{\partial v}{\partial x} -\frac{\partial p}{\partial y} + \sqrt{\frac{Pr}{Ra}}\nabla^2 v,\\
\frac{\partial w}{\partial t} &= -u_b\frac{\partial w}{\partial x} -\frac{\partial p}{\partial z} + \sqrt{\frac{Pr}{Ra}}\nabla^2 w +(\theta-\Lambda c),\\
\frac{\partial \theta}{\partial t} &= -u_b\frac{\partial \theta}{\partial x} + w + \frac{1}{\sqrt{PrRa}}\nabla^2 \theta,\\
\frac{\partial c}{\partial t} &= -u_b\frac{\partial c}{\partial x} + w + \frac{1}{Le\sqrt{PrRa}}\nabla^2 c.
\end{align}

Linear terms:
\begin{align}
u_b &= Ri^{-1/2}z,\\
\partial_z \bar{u} &= 1,\\
\partial_x &= ik_x,\\
\partial_y &= ik_y,\\
\partial_z &= D1,\\
\nabla &= \left[\begin{array}{ccc}ik_x & ik_y & \partial_z\end{array}\right],\\
\nabla^2 &= \left[\begin{array}{ccc}-k_x^2 & -k_y^2 & \partial_z^2\end{array}\right],\\
\end{align}

\begin{align}
&\left[ \begin{array}{ccccc}
1 &   &   &   &   & \\
  & 1 &   &   &   & \\
  &   & 0 &   & \\
  &   &   & 1 & \\
  &   &   &   & 1\end{array} \right] \partial t
\left[ \begin{array}{c}
u \\
v \\
p \\
\theta \\
c \end{array} \right]\\ 
&=
\left[ \begin{array}{c|c|c|c|c}
-u_b\partial x+\sqrt{\frac{Pr}{Ra}}\nabla^2 & -1  & -\partial x &  & \\
  & -u_b\partial x+\sqrt{\frac{Pr}{Ra}}\nabla^2 & -\partial y & 1 & -\Lambda\\
\partial x & \partial y &  &  &   \\
  &   &   & -& -\\
  &  &   & - & K \end{array} \right]
\left[ \begin{array}{c}
u \\
v \\
p \\
T \\
S \end{array} \right]\\
&=\left[ \begin{array}{c|c|c|c|c|c}
-u_b\partial x+\sqrt{\frac{Pr}{Ra}}\nabla^2 &  & DU & -ik_x &  & \\
 & -u_b\partial x+\sqrt{\frac{Pr}{Ra}}\nabla^2 &  & -ik_y  &  & \\
 &  & M & -D1 & G & -G\\
ik_x  & ik_y  & D1 &  &  & \\
 &  & 1 &  & N & \\
 &  & R_\rho &  &  & K\end{array} \right]
\left[ \begin{array}{c}
u \\
v \\
w \\
p \\
T \\
S \end{array} \right]
\end{align}

where
\begin{align}
M &= -u_b\partial x+\sqrt{\frac{Pr}{Ra}}\nabla^2 = -yik_x I+\sqrt{\frac{Pr}{Ra}}(-k_x^2 I +D2),\\
N &= -y\partial x+\frac{1}{\sqrt{PrRa}}\nabla^2 = -yik_x I+\frac{1}{\sqrt{PrRa}}(-k_x^2 I+D2),\\
K &= -y\partial x+\frac{1}{Le\sqrt{PrRa}}\nabla^2 = -yik_x I+\frac{1}{Le\sqrt{PrRa}}(-k_x^2 I+D2),\\
DU &= - 2\pi cos(2\pi z),\\
G &= Ri^{-1/2}.
\end{align}

Now we can compute the eigenvalue of system matrix.
\begin{align}
B &= \left[ \begin{array}{cccccc}
1 &   &   &   &   & \\
  & 1 &   &   &   & \\
  &   & 1 &   &   & \\
  &   &   & 0 &   & \\
  &   &   &   & 1 & \\
  &   &   &   &   & 1\end{array} \right],\\
A &= \left[ \begin{array}{c|c|c|c|c|c}
M &  & DU & -ik_x &  & \\
 & M &  & -ik_y  &  & \\
 &  & M & -D1 & G & -G\\
ik_x  & ik_y  & D1 &  &  & \\
 &  & 1 &  & N & \\
 &  & R_\rho &  &  & K\end{array} \right]
\end{align}

eigenvalue = $eig(A,B).$
