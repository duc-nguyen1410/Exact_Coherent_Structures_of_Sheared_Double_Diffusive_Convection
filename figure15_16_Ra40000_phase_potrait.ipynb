# Energy balance between energy input and viscous dissipation
To confirm the shear-influenced regime within the present parameter regime, we have computed the contributions of shear production and buoyant energy input to the kinetic energy balance. By taking the inner product between $\boldsymbol{U}$ and \eqref{eq.sddc_nondimensional.momentum} and then taking the volume average ($\langle \cdot \rangle$) over the entire domain, the kinetic energy equation can be determined by
$$\frac{1}{2} \frac{\partial}{\partial t} \left\langle \boldsymbol{U}^2 \right\rangle = \sqrt{\frac{Pr}{Ra}} \tau_w U_w + \left\langle (T-\Lambda S)w \right\rangle - \sqrt{\frac{Pr}{Ra}} \left\langle (\nabla \times \boldsymbol{U})^2 \right\rangle = I_w + I_b - D,$$

where $I_w$ and $I_b$ are the rates of energy input from the walls \citep{halcrow2008charting} and buoyancy forces \citep{reetz2019turbulent}. In equation \eqref{eq:energy_balance}, $D$ is the viscous dissipation, and $\tau_w$ is the wall-shear stress

$$\tau_w := 1 + \frac{1}{2L_x L_y} \int_{0}^{L_x} \int_{0}^{L_y} \left( \left.\frac{\partial u}{\partial z} \right|_{z=-0.5} + \left.\frac{\partial u}{\partial z} \right|_{z=0.5}\right) \,dx \,dy.$$

After performing a time average of \eqref{eq:energy_balance}, all equilibrium states and statistically steady states should satisfy $\overline{I}_w + \overline{I}_b = \overline{D}$

```python
from func import *
Ra = 40000.0
Pr = 7.0
Uw = 1.0
def get_energy_component_dns(path):
    [_,D]=readdata(path,4) # viscous dissipation
    [_,tau_w]=readdata(path,6) # wall-shear stress
    [_,I_b]=readdata(path,23) # bouyancy force
    I_w = np.sqrt(Pr/Ra)*tau_w*Uw # compute shear production
    return I_w,I_b,D
def get_energy_component_ecs(path):
    [_,D]=readdata(path,3) # viscous dissipation
    [_,tau_w]=readdata(path,5) # wall-shear stress
    [_,I_b]=readdata(path,22) # bouyancy force
    I_w = np.sqrt(Pr/Ra)*tau_w*Uw # compute shear production
    return I_w[-1],I_b[-1],D[-1]

I_0 = np.sqrt(Pr/Ra)*1.0*Uw


plt.figure(figsize=(5, 3))

I_w, I_b, D = get_energy_component_dns("Data/Ra40000/dns/energy.asc")
D=D/0.0025*np.sqrt(Pr/Ra) # convert to correct value, dns uses the old code that use nu=1/Re(=400)
plt.plot((I_w+I_b)/I_0,(I_w+I_b)/D, label='DNS',color='k',alpha=0.3,linewidth=1)

I_w, I_b, D = get_energy_component_dns("Data/Ra40000/PO1/energy.asc")
D=D/0.0025*np.sqrt(Pr/Ra)
plt.plot((I_w+I_b)/I_0,(I_w+I_b)/D,markersize='1.5',color='tab:brown', linewidth= '0.8', label='PO1',alpha=1,zorder=2)

I_w, I_b, D = get_energy_component_dns("Data/Ra40000/PO2/energy.asc")
D=D/0.0025*np.sqrt(Pr/Ra)
plt.plot((I_w+I_b)/I_0,(I_w+I_b)/D,markersize='1.5',color='m', linewidth= '0.8', label='PO2',alpha=1,zorder=2)

I_w, I_b, D = get_energy_component_ecs("Data/Ra40000/E1_up/fieldconverge.asc")
plt.scatter((I_w+I_b)/I_0,(I_w+I_b)/D, label='E1',marker='o',color='r',zorder=2)

I_w, I_b, D = get_energy_component_ecs("Data/Ra40000/E1_low/fieldconverge.asc")
plt.scatter((I_w+I_b)/I_0,(I_w+I_b)/D, label='C12',marker='o',facecolors='none',color='r',zorder=2)

I_w, I_b, D = get_energy_component_ecs("Data/Ra40000/E2_up/fieldconverge.asc")
plt.scatter((I_w+I_b)/I_0,(I_w+I_b)/D, label='E2$_{up}$',marker='s',color='g',zorder=2)

I_w, I_b, D = get_energy_component_ecs("Data/Ra40000/E2_low/fieldconverge.asc")
plt.scatter((I_w+I_b)/I_0,(I_w+I_b)/D, label='E2$_{low}$',marker='s',facecolors='none',color='g',zorder=2)

I_w, I_b, D = get_energy_component_ecs("Data/Ra40000/E3_up/fieldconverge.asc")
plt.scatter((I_w+I_b)/I_0,(I_w+I_b)/D, label=r'E3$_{up}$',marker='v',color='k',zorder=2)

I_w, I_b, D = get_energy_component_ecs("Data/Ra40000/E3_low/fieldconverge.asc")
plt.scatter((I_w+I_b)/I_0,(I_w+I_b)/D, label=r'E3$_{low}$',marker='v',facecolors='none',color='k',zorder=2)

I_w, I_b, D = get_energy_component_ecs("Data/Ra40000/E4_up/fieldconverge.asc")
plt.scatter((I_w+I_b)/I_0,(I_w+I_b)/D, label=r'E4$_{up}$',marker='D',color='tab:orange',zorder=2)

I_w, I_b, D = get_energy_component_ecs("Data/Ra40000/E4_low/fieldconverge.asc")
plt.scatter((I_w+I_b)/I_0,(I_w+I_b)/D, label=r'E4$_{low}$',marker='D',facecolors='none',color='tab:orange',zorder=2)

plt.scatter(1,1, label='Base',marker='.',color='b',zorder=2)


plt.xlabel(r'$(I_w+I_b)/I_0$', fontsize=14)
plt.ylabel(r'$(I_w+I_b)/D$', fontsize=14)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
# plt.xticks([1,1.05,1.1,1.15],['1','1.05','1.1','1.15'],fontsize=14)
# plt.yticks([0,0.25,0.5,0.75,1],['0','0.25','0.5','0.75','1'],fontsize=14)
# plt.ylim(top=1.1)

# xlims = plt.xlim()
# ylims = plt.ylim()
# minlim = min(xlims[0], ylims[0])
# maxlim = max(xlims[1], ylims[1])
# plt.plot([minlim, maxlim], [minlim, maxlim], '--', color='tab:blue', label=r'$I_w+I_b = D$', zorder=1)
plt.gca().axhline(y=1, color='k', zorder=0, linewidth=0.5) #, label=r'$I_w+I_b = D$'
plt.gca().axvline(x=1, color='k', zorder=0, linewidth=0.5)
# plt.xlim(xlims)
# plt.ylim(ylims)

plt.legend(
        loc='right', 
        bbox_to_anchor=(1.25, 0.5),
        ncol=1, 
        frameon=False
    )

plt.savefig('Data/Ra40000/energy_I_D.png', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```
