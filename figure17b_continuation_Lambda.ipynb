## Plot bifurcation diagram with stability
- Solid line: stable
- Dashed line: unstable (real part > $10^{-6}$)


```python
from func import *

fontsize = 16
plt.figure(figsize=(5, 3))

print("get E1 bifurcation")
Lambda, Nu, instability = np.load("Data/Equilibrium_parameter_effect/continuation_E1_Lambda_Nu.npz").values()
segments = separate_into_segments(Lambda, Nu, instability)
plot_stability_bifurcation(segments,'E1','r')


print("get E2 bifurcation")
Lambda, Nu, instability = np.load("Data/Equilibrium_parameter_effect/continuation_E2_Lambda_Nu.npz").values()
segments = separate_into_segments(Lambda, Nu, instability)
plot_stability_bifurcation(segments,'E2','g')

print("get E3 bifurcation")
Lambda, Nu, instability = np.load("Data/Equilibrium_parameter_effect/continuation_E3_Lambda_Nu.npz").values()
segments = separate_into_segments(Lambda, Nu, instability)
plot_stability_bifurcation(segments,'E3','k')

plt.xlabel(r'$\Lambda$', fontsize=fontsize)
plt.ylabel(r'$Nu$', fontsize=fontsize)
plt.xticks(fontsize=fontsize)
plt.yticks(fontsize=fontsize)
# plt.ylim(bottom=1)
plt.xlim(left=0)
plt.xscale('linear')
plt.legend(frameon=False)
plt.savefig('Data/Equilibrium_parameter_effect/EQCont_LambdaNu.png', dpi=300, bbox_inches='tight')
plt.show()
plt.close()
```
