# Plot phase portrait (Jt, Nu) of DNS at $Ra=4\times 10^4$ visiting all ECS

```python
from func import *
from matplotlib.animation import FuncAnimation
from matplotlib.gridspec import GridSpec
plt.rcParams['font.size'] = 14


fig = plt.figure(figsize=(10, 5))
gs = GridSpec(2, 2, figure=fig, height_ratios=[3, 1.5])

ax1 = fig.add_subplot(gs[0,0])  # row 0 column 0, Jt-Nu
ax2 = fig.add_subplot(gs[0,1])  # row 0 column 1, Nu
ax3 = fig.add_subplot(gs[1,0])  # row 1 column 0, Temperature
ax4 = fig.add_subplot(gs[1,1])  # row 1 column 1, salinity

tDNS, JtDNS = readdata("../2D/highPr/Pr7/Ra40000/dns_cont/energy.asc", 16)
tDNS, NuDNS = readdata("../2D/highPr/Pr7/Ra40000/dns_cont/energy.asc", 17)

################################
##### row 0 column 0, Jt-Nu
x_idx = 15
y_idx = 16
start=0
[_,Jt]=readdata("Data/Ra40000/dns/energy.asc",x_idx+1)
[_,Nu]=readdata("Data/Ra40000/dns/energy.asc",y_idx+1)
ax1.plot(Jt[start:],Nu[start:], label='DNS',color='k',alpha=0.3,linewidth=1)

[_,JtPO1]=readdata("Data/Ra40000/PO1/energy.asc",x_idx+1)
[_,NuPO1]=readdata("Data/Ra40000/PO1/energy.asc",y_idx+1)
ax1.plot(JtPO1,NuPO1,'-',markersize='1.5',color='tab:brown', linewidth= '0.8', label='PO1',alpha=1,zorder=2)

[_,JtPO2]=readdata("Data/Ra40000/PO2/energy.asc",x_idx+1)
[_,NuPO2]=readdata("Data/Ra40000/PO2/energy.asc",y_idx+1)
ax1.plot(JtPO2,NuPO2,'-',markersize='1.5',color='m', linewidth= '0.8', label='PO2',alpha=1,zorder=2)

[_,Jt]=readdata("Data/Ra40000/E1_up/fieldconverge.asc",x_idx)
[_,Nu]=readdata("Data/Ra40000/E1_up/fieldconverge.asc",y_idx)
ax1.scatter(Jt[-1],Nu[-1], label='E1',marker='o',color='r',zorder=2)

[_,Jt]=readdata("Data/Ra40000/E1_low/fieldconverge.asc",x_idx)
[_,Nu]=readdata("Data/Ra40000/E1_low/fieldconverge.asc",y_idx)
ax1.scatter(Jt[-1],Nu[-1], label='C12',marker='o',facecolors='none',color='r',zorder=2)

[_,Jt]=readdata("Data/Ra40000/E2_up/fieldconverge.asc",x_idx)
[_,Nu]=readdata("Data/Ra40000/E2_up/fieldconverge.asc",y_idx)
ax1.scatter(Jt[-1],Nu[-1], label='E2$_{up}$',marker='s',color='g',zorder=2)

[_,Jt]=readdata("Data/Ra40000/E2_low/fieldconverge.asc",x_idx)
[_,Nu]=readdata("Data/Ra40000/E2_low/fieldconverge.asc",y_idx)
ax1.scatter(Jt[-1],Nu[-1], label='E2$_{low}$',marker='s',facecolors='none',color='g',zorder=2)

[_,Jt]=readdata("Data/Ra40000/E3_up/fieldconverge.asc",x_idx)
[_,Nu]=readdata("Data/Ra40000/E3_up/fieldconverge.asc",y_idx)
ax1.scatter(Jt[-1],Nu[-1], label=r'E3$_{up}$',marker='v',color='k',zorder=2)

[_,Jt]=readdata("Data/Ra40000/E3_low/fieldconverge.asc",x_idx)
[_,Nu]=readdata("Data/Ra40000/E3_low/fieldconverge.asc",y_idx)
ax1.scatter(Jt[-1],Nu[-1], label=r'E3$_{low}$',marker='v',facecolors='none',color='k',zorder=2)

[_,Jt]=readdata("Data/Ra40000/E4_up/fieldconverge.asc",x_idx)
[_,Nu]=readdata("Data/Ra40000/E4_up/fieldconverge.asc",y_idx)
ax1.scatter(Jt[-1],Nu[-1], label=r'E4$_{up}$',marker='D',color='tab:orange',zorder=2)

[_,Jt]=readdata("Data/Ra40000/E4_low/fieldconverge.asc",x_idx)
[_,Nu]=readdata("Data/Ra40000/E4_low/fieldconverge.asc",y_idx)
ax1.scatter(Jt[-1],Nu[-1], label=r'E4$_{low}$',marker='D',facecolors='none',color='tab:orange',zorder=2)

ax1.scatter(1,1, label='Base',marker='.',color='b',zorder=2)

ax1.set_xlabel(r'$J_T$', fontsize=14)
ax1.set_ylabel(r'$Nu$', fontsize=14)
ax1.set_ylim([0.4,2.9])
xlims = ax1.get_xlim()
ylims = ax1.get_ylim()
minlim = min(xlims[0], ylims[0])
maxlim = max(xlims[1], ylims[1])
ax1.plot([minlim, maxlim], [minlim, maxlim], '--', color='tab:blue', label=r'$Nu = J_T$', zorder=1)
ax1.set_xlim(xlims)
ax1.set_ylim(ylims)
p1 = ax1.scatter([], [], marker='o', color='tab:brown', s=20, label='PO1')
p2 = ax1.scatter([], [], marker='o', color='m', s=20, label='PO2')
p3 = ax1.scatter([], [], marker='o', color='k', s=20)


################################
##### row 0 column 1, Nu: from t=6300 to ~6700
# [time,Nu] = readdata("Data/Ra40000/dns/energy.asc",17)
# ax2.plot(time[0:],Nu[0:],'-',color='k')
ax2.plot(tDNS[0:],NuDNS[0:],'-',color='k')
[_,Nu]=readdata("Data/Ra40000/E1_up/fieldconverge.asc",16)
ax2.axhline(y=Nu[-1], color='r', label='E1', zorder=2)
[_,Nu]=readdata("Data/Ra40000/E2_up/fieldconverge.asc",16)
ax2.axhline(y=Nu[-1], color='g', label='E2', zorder=2)

ax2.set_xlabel(r'$t$',fontsize=14)
ax2.set_ylabel(r'$Nu$',fontsize=14)

ax2.set_xlim([tDNS[0],tDNS[400]])
ax2.set_ylim([0.4,2.9])
ax2.tick_params(axis='x', which='both', bottom=True, top=False, labelbottom=True)
ax2.tick_params(axis='y', which='both', left=True, right=False, labelleft=True)

ax2.axvspan(6350, 6410, color='tab:blue',edgecolor='tab:blue', linewidth=0,alpha=0.2, zorder=0) # E2
ax2.text(6380,0.95*ax2.get_ylim()[1],'E2',ha='center', va='top', fontsize=14)
ax2.axvspan(6410, 6470, color='tab:orange',edgecolor='tab:orange', linewidth=0,alpha=0.2, zorder=0) # E1
ax2.text(6440,0.95*ax2.get_ylim()[1],'E1',ha='center', va='top', fontsize=14)

ax2.axvspan(6530, 6600, color='tab:blue',edgecolor='tab:blue', linewidth=0,alpha=0.2, zorder=0) # E2
ax2.text(6565,0.95*ax2.get_ylim()[1],'E2',ha='center', va='top', fontsize=14)
ax2.axvspan(6600, 6670, color='tab:orange',edgecolor='tab:orange', linewidth=0,alpha=0.2, zorder=0) # E1
ax2.text(6635,0.95*ax2.get_ylim()[1],'E1',ha='center', va='top', fontsize=14)

p4 = ax2.scatter([], [], marker='o', color='k', s=20)

################################
##### row 1 column 0, Temperature
[zcoord,xcoord,temp] = z_averaged_scalar('../2D/highPr/Pr7/Ra40000/dns_cont/t6300.nc',2)
X_ax3, Z_ax3 = np.meshgrid(xcoord, zcoord)
p5 = ax3.pcolormesh(X_ax3,Z_ax3,(temp-zcoord).T, shading='auto', cmap='bwr')
ax3.set_xlabel(r'$x$', fontsize=14)
ax3.set_ylabel(r'$z$', fontsize=14)
ax3.set_xlim(0, 2*np.pi)
ax3.set_ylim(-0.5, 0.5)
plt.colorbar(p5, ax=ax3,orientation='horizontal',location='top', shrink=1.0, aspect=10, label=r'$\theta$')

################################
##### row 1 column 1, Salinity
[zcoord,xcoord,salt] = z_averaged_scalar('../2D/highPr/Pr7/Ra40000/dns_cont/s6300.nc',2)
X_ax4, Z_ax4 = np.meshgrid(xcoord, zcoord)
p6 = ax4.pcolormesh(X_ax4,Z_ax4,(salt-zcoord).T, shading='auto', cmap='bwr')
ax4.set_xlabel(r'$x$', fontsize=14)
ax4.set_ylabel(r'$z$', fontsize=14)
ax4.set_xlim(0, 2*np.pi)
ax4.set_ylim(-0.5, 0.5)
plt.colorbar(p6, ax=ax4,orientation='horizontal',location='top', shrink=1.0, aspect=10, label=r'$s$')


plt.tight_layout()

step = 2
nframes = int(len(NuDNS)/step) # number of frames in the animation
# nframes = 2 # for testing
print("nframes = ",nframes)

# --- Update function ---
def update(frame):
    # To avoid index errors, loop within orbit length
    i1 = frame % len(NuPO1)
    i2 = frame % len(NuPO2)
    i3 = frame * step
    p1.set_offsets([[JtPO1[i1], NuPO1[i1]]])
    p2.set_offsets([[JtPO2[i2], NuPO2[i2]]])
    p3.set_offsets([[JtDNS[i3], NuDNS[i3]]])
    p4.set_offsets([[tDNS[i3], NuDNS[i3]]])

    start_index = 6300
    index = start_index + i3
    print(f"Updating frame {frame}, index {index}")

    [zcoord,_,temp] = z_averaged_scalar(f'../2D/highPr/Pr7/Ra40000/dns_cont/t{index}.nc',2)
    [zcoord,_,salt] = z_averaged_scalar(f'../2D/highPr/Pr7/Ra40000/dns_cont/s{index}.nc',2)
    p5.set_array((temp - zcoord).T)
    p6.set_array((salt - zcoord).T)
    return p1, p2, p3, p4, p5, p6


# # --- Create animation ---
ani = FuncAnimation(fig, update, frames=nframes, interval=40, blit=False, repeat=True)
ani.save('Supplementary_movie_2.mp4', writer='ffmpeg', fps=30, dpi=200, bitrate=2000)


plt.show()
plt.close()
```
