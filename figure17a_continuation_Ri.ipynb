## Plot bifurcation diagram with stability

- Dashed line: unstable (real part > $10^{-6}$)


```python
from func import *

fontsize = 16
plt.figure(figsize=(5, 3))

print("get E1 bifurcation")
Ri, Nu, instability = np.load("Data/Equilibrium_parameter_effect/continuation_E1_Ri_Nu.npz").values()
segments = separate_into_segments(Ri, Nu, instability)
plot_stability_bifurcation(segments,'E1','r')
plt.scatter(Ri[50],Nu[50],marker='d',facecolors='none',color='r',zorder=3)

print("get E2 bifurcation")
Ri, Nu, instability = np.load("Data/Equilibrium_parameter_effect/continuation_E2_Ri_Nu.npz").values()
segments = separate_into_segments(Ri, Nu, instability)
plot_stability_bifurcation(segments,'E2','g')
plt.scatter(Ri[38],Nu[38],marker='d',facecolors='none',color='g',zorder=3)

print("get E3 bifurcation")
Ri, Nu, instability = np.load("Data/Equilibrium_parameter_effect/continuation_E3_Ri_Nu.npz").values()
segments = separate_into_segments(Ri, Nu, instability)
plot_stability_bifurcation(segments,'E3','k')

plt.scatter([],[],marker='d',facecolors='none',label='HB',color='k',zorder=3)

plt.xlabel(r'$Ri$', fontsize=fontsize)
plt.ylabel(r'$Nu$', fontsize=fontsize)
plt.xticks(fontsize=fontsize)
plt.yticks(fontsize=fontsize)
plt.xlim([0,3])
plt.xscale('linear')
plt.legend(frameon=False)
plt.savefig('Data/Equilibrium_parameter_effect/EQCont_RiNu.png', dpi=300, bbox_inches='tight')

plt.show()
plt.close()
```
